```python
def hamming_encode(data):
    m = len(data)
    r = 0

    while 2 ** r < m + r + 1:
        r += 1

    n = m + r
    code = ['0'] * (n + 1)

    j = 0
    for i in range(1, n + 1):
        if i & (i - 1):
            code[i] = data[j]
            j += 1

    print(f"\nАқпараттық биттер: {m}")
    print(f"Бақылау биттері: {r}")
    print(f"Жалпы биттер: {n}")
    print("Бақылау позициялары:", [2 ** i for i in range(r)])

    for k in range(r):
        p = 2 ** k
        positions = [i for i in range(1, n + 1) if i & p]
        value = sum(int(code[i]) for i in positions) % 2
        code[p] = str(value)

        print(f"a{p}: {positions} → {value}")

    return ''.join(code[1:])


def hamming_decode(received):
    n = len(received)
    code = ['0'] + list(received)
    syndrome = 0

    for k in range(n.bit_length()):
        p = 2 ** k
        positions = [i for i in range(1, n + 1) if i & p]
        value = sum(int(code[i]) for i in positions) % 2

        print(f"E{k + 1}: {positions} → {value}")

        if value:
            syndrome += p

    print(f"\nСиндром: {syndrome:b} (екілік)")
    print(f"Синдром: {syndrome} (ондық)")

    if syndrome == 0:
        print("Қате анықталмады.")
    elif syndrome <= n:
        print(f"Қате {syndrome}-позицияда.")
        code[syndrome] = str(1 - int(code[syndrome]))
    else:
        print("Синдром код ұзындығынан үлкен. Түзету мүмкін емес.")
        return

    print("Түзетілген код:", ''.join(code[1:]))


while True:
    data = input("\nЕкілік сан енгізіңіз: ").strip()

    if data and all(bit in '01' for bit in data):
        break

    print("Қате! Тек 0 және 1 енгізіңіз.")

encoded = hamming_encode(data)
print("\nДұрыс Хэмминг коды:", encoded)

while True:
    try:
        error = int(input("\nҚате болатын позиция (0 - қатесіз): "))
        if 0 <= error <= len(encoded):
            break
    except ValueError:
        pass

    print("Позицияны дұрыс енгізіңіз!")

received = list(encoded)

if error:
    received[error - 1] = str(1 - int(received[error - 1]))

received = ''.join(received)

print("\nҚабылданған код:", received)
hamming_decode(received)
```
