```python
def uniform_code(n):
    k = (n - 1).bit_length()

    print("\n1-ТАПСЫРМА")
    print("Алфавит қуаты N =", n)
    print("Минималды бит саны k =", k)
    print(f"2^{k} = {2 ** k} >= {n}")


def check_fano(codes):
    values = list(codes.values())

    for i in range(len(values)):
        for j in range(len(values)):
            if i != j and values[j].startswith(values[i]):
                return False

    return True


def decode(message, codes):
    reverse = {code: symbol for symbol, code in codes.items()}
    result = []
    parts = []
    current = ""

    for bit in message:
        current += bit

        if current in reverse:
            result.append(reverse[current])
            parts.append(current)
            current = ""

    if current:
        print("Қате: хабар толық декодталмайды!")
        return

    print("\n2-ТАПСЫРМА")
    print("Бастапқы хабар:", message)
    print("Кодтарға бөлінуі:", " | ".join(parts))
    print("Декодталған хабар:", "".join(result))


while True:
    try:
        n = int(input("Алфавиттегі символдар саны N: "))
        if n > 0:
            break
    except ValueError:
        pass
    print("Оң бүтін сан енгізіңіз!")

uniform_code(n)

while True:
    try:
        count = int(input("\nКодталатын символдар саны: "))
        if count > 0:
            break
    except ValueError:
        pass
    print("Оң бүтін сан енгізіңіз!")

while True:
    codes = {}

    for i in range(count):
        symbol = input(f"{i + 1}-символ: ").strip()

        while not symbol or symbol in codes:
            symbol = input("Басқа символ енгізіңіз: ").strip()

        code = input(f"{symbol} символының коды: ").strip()

        while not code or any(bit not in "01" for bit in code):
            code = input("Тек 0 және 1 енгізіңіз: ").strip()

        codes[symbol] = code

    if check_fano(codes):
        print("\nФано шарты орындалады!")
        break

    print("\nФано шарты орындалмайды!")
    print("Кодтарды қайта енгізіңіз.\n")

while True:
    message = input("\nЕкілік хабарды енгізіңіз: ").strip()

    if message and all(bit in "01" for bit in message):
        break

    print("Тек 0 және 1 енгізіңіз!")

decode(message, codes)
```
